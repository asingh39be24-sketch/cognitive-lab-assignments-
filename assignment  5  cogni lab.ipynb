```python
import numpy as np

# Q1
arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)
print("a) Add 2:", arr + 2)
print("b) Multiply by 3:", arr * 3)
print("c) Divide by 2:", arr / 2)
```

```python
# Q2 a
arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed array:", arr[::-1])
```

```python
# Q2 b
def most_frequent(a):
    values, counts = np.unique(a, return_counts=True)
    for v in values[counts == counts.max()]:
        print("Most frequent value:", v, "| Count:", counts.max(),
              "| Indices:", np.where(a == v)[0])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

print("i. Array x")
most_frequent(x)
print("ii. Array y")
most_frequent(y)
```

```python
# Q3
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("a) 1st row, 2nd column:", arr[0, 1])
print("b) 3rd row, 1st column:", arr[2, 0])
```

```python
# Q4
aryan = np.linspace(10, 100, 25)
print(aryan)
print("Dimensions:", aryan.ndim)
print("Shape:", aryan.shape)
print("Total elements:", aryan.size)
print("Data type:", aryan.dtype)
print("Total bytes:", aryan.nbytes)
```

```python
# Q5
ucs420_aryan = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])
print(ucs420_aryan)
print("Mean:", np.mean(ucs420_aryan))
print("Median:", np.median(ucs420_aryan))
print("Max:", np.max(ucs420_aryan))
print("Min:", np.min(ucs420_aryan))
print("Unique elements:", np.unique(ucs420_aryan))
```

```python
# Q5 reshape and resize
reshaped_ucs420_aryan = ucs420_aryan.reshape(4, 3)
print("Reshaped to 4 x 3:")
print(reshaped_ucs420_aryan)

resized_ucs420_aryan = np.resize(ucs420_aryan, (2, 3))
print("Resized to 2 x 3:")
print(resized_ucs420_aryan)
```
