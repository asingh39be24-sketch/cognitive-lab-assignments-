```python
# Create the tuple
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)
print("Scores:", scores)

# i. 
highest = max(scores)
print("Highest score:", highest, "| Index:", scores.index(highest))

# ii. 
lowest = min(scores)
print("Lowest score:", lowest, "| Count:", scores.count(lowest))

# iii.
reversed_list = list(reversed(scores))
print("Reversed as list:", reversed_list)

# iv.
target = float(input("Enter a score to search: "))
if target in scores:
    print("Found at index:", scores.index(target))
else:
    print("Score not present in the tuple")
```

```python
import random

# Create list of 100 random numbers between 100 and 900
nums = [random.randint(100, 900) for _ in range(100)]
print("Random numbers:", nums)

# Prime check
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# i. Odd numbers
odd = [n for n in nums if n % 2 != 0]
print("\nOdd numbers:", odd)
print("Count of odd numbers:", len(odd))

# ii. Even numbers
even = [n for n in nums if n % 2 == 0]
print("\nEven numbers:", even)
print("Count of even numbers:", len(even))

# iii. Prime numbers
prime = [n for n in nums if is_prime(n)]
print("\nPrime numbers:", prime)
print("Count of prime numbers:", len(prime))
```

```python
# Create the sets
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}
print("Set A:", A)
print("Set B:", B)

# i. Union (unique scores of both teams)
print("Union:", A.union(B))

# ii. Intersection (common scores)
print("Intersection:", A.intersection(B))

# iii. Symmetric difference (exclusive to each team)
print("Symmetric difference:", A.symmetric_difference(B))

# iv. Subset and superset check
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# v. Remove a score X from A if it exists
X = int(input("Enter score X to remove from A: "))
if X in A:
    A.remove(X)
    print("Removed", X, "| Updated A:", A)
else:
    print(X, "is not present in set A")
```
