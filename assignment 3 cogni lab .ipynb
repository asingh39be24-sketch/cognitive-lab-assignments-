```python
# Q6
employees = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
pd.DataFrame(employees).to_csv("employees.csv", index=False)
emp = pd.read_csv("employees.csv")
emp
```

```python
# Q6 a
print("Shape:", emp.shape)
print("Rows:", emp.shape[0])
print("Columns:", emp.shape[1])
```

```python
# Q6 e
print("Average salary:", emp["Salary"].mean())
print("Total bonus paid:", emp["Bonus"].sum())
print("Youngest employee's age:", emp["Age"].min())
print("Highest performance rating:", emp["Rating"].max())
```

```python
# Q6 f
emp.sort_values(by="Salary", ascending=False)
```

```python
# Q6 g
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp["Performance"] = emp["Rating"].apply(categorize)
emp
```

```python
# Q6 h
print(emp.isnull().sum())
print("Total missing values:", emp.isnull().sum().sum())
```
