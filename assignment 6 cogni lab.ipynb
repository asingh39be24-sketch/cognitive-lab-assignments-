```python
import numpy as np

# Q1
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected = temperature + 2
print("a) After adding 2:", corrected)

fahrenheit = (9 / 5) * temperature + 32
print("b) In Fahrenheit:", fahrenheit)

print("c) Readings above 32:", temperature[temperature > 32])
print("d) Count above 32:", np.sum(temperature > 32))
```

```python
# Q2
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])

print("a) Total steps:", steps.sum())
print("b) Mean steps:", steps.mean())
print("c) Maximum:", steps.max(), "| Minimum:", steps.min())
print("d) Total per day (axis=0):", steps.sum(axis=0))
print("e) Total per user (axis=1):", steps.sum(axis=1))

user, day = np.unravel_index(steps.argmax(), steps.shape)
print("f) argmax (flat index):", steps.argmax())
print("   Maximum is at row", user, "column", day, "-> User", user + 1, "Day", day + 1)
```

```python
# Q4 a, b
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]])
y = np.array([40, 65, 30, 85])

print("a) Shape:", X.shape, "| Dimensions:", X.ndim)
print("b) X.T:")
print(X.T)
```

```python
# Q4 c, d
XtX = X.T @ X
print("c) X.T @ X:")
print(XtX)

XtX_inv = np.linalg.inv(XtX)
print("d) Inverse of X.T @ X:")
print(XtX_inv)
```

```python
# Q4 e, g
beta = XtX_inv @ X.T @ y
print("e) Coefficients (beta):", beta)

new_user = np.array([5, 40, 7])
print("g) Predicted assistance score:", new_user @ beta)
```
